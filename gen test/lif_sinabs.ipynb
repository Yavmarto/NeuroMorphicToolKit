```python
"""Sinabs model definitions — auto-generated from NIR."""

import numpy as np
import torch
import torch.nn as nn
import sinabs.layers as sl
from sinabs.network import Network as SinabsNetwork

def build_model():
    layer_0 = nn.Linear(in_features=1, out_features=1, bias=False)
    with torch.no_grad():
        layer_0.weight.copy_(torch.tensor([[1.0]], dtype=torch.float32))
    layer_1 = sl.LIF(tau_mem=0.0025)
    model = nn.Sequential(layer_0, layer_1)
    return SinabsNetwork(model)
```
